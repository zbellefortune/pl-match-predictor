```python
import pandas as pd
res = []
season = ["1920", "2021", "2122", "2223", "2324", "2425", "2526"]
for i in season:
    # read the file
    dfcur = pd.read_csv(f"../data/raw/E0_{i}.csv")
    # add season column
    dfcur["Season"] = i
    # put res on a list
    res.append(dfcur)
    #print(dfcur.shape)

df = pd.concat(res, ignore_index = True)
df.head()
df.isna().sum()
pd.set_option("display.max_rows", None)
#cleaning the data
df = df.loc[:, ["Date", "Season", "HomeTeam", "AwayTeam", "FTR", "FTAG"]]
df.head()
```
